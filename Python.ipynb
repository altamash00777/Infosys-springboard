```python
#check prime
```

```python
n = 29

if n < 2:
    print("Not Prime")
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```

```python
# Print Prime Numbers from 1 to 100
```

```python
for n in range(2, 101):
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            break
    else:
        print(n, end=" ")
```

```python
#Reverse integer
```

```python
n = 12345
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n //= 10

print(reverse)
```

```python
#sum of digit
```

```python
n = 12345
total = 0

while n > 0:
    total += n % 10
    n //= 10

print(total)
```
